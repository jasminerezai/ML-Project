```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
import seaborn as sns

current_dir = Path.cwd().resolve()
project_root = current_dir.parent if current_dir.name == 'notebooks' else current_dir
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.preprocess import load_and_preprocess_data

data_path = project_root / 'data' / 'iris_extended.csv'
df, X_train, X_test, y_train, y_test, encoder = load_and_preprocess_data(data_path)
print(f'Dataset shape: {df.shape}')
print(f'Train set: {X_train.shape}, Test set: {X_test.shape}')

feature_columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
grid = sns.pairplot(df, vars=feature_columns, hue='species', corner=True)
grid.fig.suptitle('Iris Measurement Features by Species', y=1.02)
figure_path = project_root / 'outputs' / 'figures' / 'iris_pairplot.png'
figure_path.parent.mkdir(parents=True, exist_ok=True)
grid.fig.savefig(figure_path, bbox_inches='tight')
plt.close(grid.fig)
```
